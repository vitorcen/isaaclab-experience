## 7. 训练数据 / BEHAVIOR Challenge 2025

> ⚠️ **B-1K 基础包不带训练 trajectories** —— `behavior-1k-assets` 只有场景 / 物体 / 任务定义（BDDL）。要训 Imitation Learning policy，需要单独下 Challenge 的示范数据。

### 7a. 官方公开数据集（HuggingFace）

| 数据集 | 大小 | 内容 | URL |
|--------|------|------|-----|
| **2025-challenge-task-instances** | ~MB | 预采样好的任务场景配置（50 任务的训/测 instance），跑 cached BehaviorTask env 必需 | `github.com/StanfordVL/2025-challenge-task-instances` |
| **2025-challenge-demos** | ~1.5 TB | 10000 条人类遥操作示范，50 任务，**LeRobot 格式**（与 `lerobot/` 子模块直接对接） | `huggingface.co/datasets/behavior-1k/2025-challenge-demos` |
| **2025-challenge-rawdata** | (HDF5) | 原始 HDF5，可用 `OmniGibson/scripts/replay_obs.py` 在 OG 里精确重放、补采视觉模态 | `huggingface.co/datasets/behavior-1k/2025-challenge-rawdata` |

### 7b. 任务实例 vs 示范

- **task-instances** = 「初始场景配置」（哪些物体放哪儿）—— 几 MB，所有评测都需要
- **demos** = 「人类成功完成任务的 rollout」 —— 1.5 TB，IL 训练才需要

### 7c. 按需下载（不用真下 1.5 TB）

`2025-challenge-demos` 支持**按 task / 相机 / 模态部分下载**（官方文档原话）。只关心某几个任务就下那部分，几个 GB 起步。教程：

- `github.com/StanfordVL/b1k-baselines/blob/main/tutorials/dataset.ipynb` —— 加载数据集
- `github.com/StanfordVL/b1k-baselines/blob/main/tutorials/generate_custom_data.ipynb` —— 生成新模态

### 7d. 官方基线 (b1k-baselines)

独立 repo：`github.com/StanfordVL/b1k-baselines`

| 类型 | 模型 |
|------|------|
| BC 基线 | ACT、Diffusion Policy、BC-RNN、WB-VIMA |
| 预训练 VLA | OpenVLA、π0 |

### 7e. 端到端训练流程

```
1. 下 task-instances        (~MB)   → 场景配置
2. 部分下 challenge-demos   (~GB)   → 选 N 个任务的 demos
3. 用 lerobot 训 ACT/DP/π0  → policy.pth
4. omnigibson.learning.eval  → 在 instance 上评测
```

### 7f. 自己采数据 (JoyLo)

`BEHAVIOR-1K/joylo/` —— 用 GELLO 设备 + Nintendo JoyCon 遥操作收 demonstration，输出 HDF5。详见 `BEHAVIOR-1K/docs/behavior_components/data_collection.md`。

你的 repo 里 `LeIsaac/` 是另一条 IL 路线（基于 Isaac Lab + lerobot，用 SO-100 / GELLO 收数据），跟 B-1K JoyLo 思路类似但场景栈不同。
